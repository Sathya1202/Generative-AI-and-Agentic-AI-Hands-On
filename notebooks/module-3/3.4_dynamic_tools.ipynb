```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.tools import tool; from typing import Dict, Any; from tavily import TavilyClient (and 1 more import statement(s)) Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Connects to or queries a relational database. Validate/limit generated SQL and keep database credentials and sensitive data out of source control.

from langchain.tools import tool
from typing import Dict, Any
from tavily import TavilyClient
from langchain_community.utilities import SQLDatabase

tavily_client = TavilyClient()

db = SQLDatabase.from_uri("sqlite:///resources/Chinook.db")


@tool
def web_search(query: str) -> Dict[str, Any]:

    """Search the web for information"""

    return tavily_client.search(query)

@tool
def sql_query(query: str) -> str:

    """Obtain information from the database using SQL queries"""

    try:
        return db.run(query)
    except Exception as e:
        return f"Error: {e}"
```

```python
# Purpose: Imports the libraries and APIs needed by the following example: from dataclasses import dataclass.

from dataclasses import dataclass

@dataclass
class UserRole:
    user_role: str = "external"
```

### Implementation step 4

Imports the libraries and APIs needed by the following example: from langchain.agents.middleware import wrap_model_call, ModelRequest, ModelResponse; from typing import Callable. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Uses runtime context to access request-scoped dependencies or contextual values without treating them as model-authored input. Registers middleware that adapts model calls, prompts or tool availability at runtime. Keep the policy deterministic and test each branch.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.agents.middleware import wrap_model_call, ModelRequest, ModelResponse; from typing import Callable. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Uses runtime context to access request-scoped dependencies or contextual values without treating them as model-authored input. Registers middleware that adapts model calls, prompts or tool availability at runtime. Keep the policy deterministic and test each branch.

from langchain.agents.middleware import wrap_model_call, ModelRequest, ModelResponse
from typing import Callable

@wrap_model_call
def dynamic_tool_call(request: ModelRequest, 
handler: Callable[[ModelRequest], ModelResponse]) -> ModelResponse:

    """Dynamically call tools based on the runtime context"""

    user_role = request.runtime.context.user_role
    
    if user_role == "internal":
        pass # internal users get access to all tools
    else:
        tools = [web_search] # external users only get access to web search
        request = request.override(tools=tools) 

    return handler(request)
```

```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.agents import create_agent. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Connects to or queries a relational database. Validate/limit generated SQL and keep database credentials and sensitive data out of source control.

from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano",
    tools=[web_search, sql_query],
    middleware=[dynamic_tool_call],
    context_schema=UserRole
)
```
