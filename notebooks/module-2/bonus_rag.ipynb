## RAG Agent

### Implementation step 11

Imports the libraries and APIs needed by the following example: from langchain.tools import tool. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Implements a RAG stage: document ingestion, chunking, embedding/indexing or similarity retrieval. Retrieved passages should be treated as evidence, not instructions.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.tools import tool. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Implements a RAG stage: document ingestion, chunking, embedding/indexing or similarity retrieval. Retrieved passages should be treated as evidence, not instructions.

from langchain.tools import tool

@tool
def search_handbook(query: str) -> str:
    """Search the employee handbook for information"""
    results = vector_store.similarity_search(query)
    return results[0].page_content
```

### Implementation step 12

Imports the libraries and APIs needed by the following example: from langchain.agents import create_agent. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.agents import create_agent. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example.

from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano",
    tools=[search_handbook],
    system_prompt="You are a helpful agent that can search the employee handbook for information."
    )
```

```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.messages import HumanMessage. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response.

from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="How many days of vacation does an employee get in their first year?")]}
)
```
