```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.tools import tool. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Connects to or queries a relational database. Validate/limit generated SQL and keep database credentials and sensitive data out of source control.

from langchain.tools import tool

@tool
def sql_query(query: str) -> str:

    """Obtain information from the database using SQL queries"""

    try:
        return db.run(query)
    except Exception as e:
        return f"Error: {e}"

sql_query.invoke("SELECT * FROM Artist LIMIT 10")
```

```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.agents import create_agent. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Connects to or queries a relational database. Validate/limit generated SQL and keep database credentials and sensitive data out of source control.

from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano",
    tools=[sql_query]
)
```

```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.messages import HumanMessage. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response.

from langchain.messages import HumanMessage

question = HumanMessage(content="Who is the most popular artist beginning with 'S' in this database?")

response = agent.invoke(
    {"messages": [question]}
)
```
