## Add web search tool

### Implementation step 7

Imports the libraries and APIs needed by the following example: from langchain.tools import tool; from typing import Dict, Any; from tavily import TavilyClient. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.tools import tool; from typing import Dict, Any; from tavily import TavilyClient. Defines a reusable Python function/tool. The docstring and type annotations communicate the tool purpose and input schema to the agent. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes.

from langchain.tools import tool
from typing import Dict, Any
from tavily import TavilyClient

tavily_client = TavilyClient()

@tool
def web_search(query: str) -> Dict[str, Any]:

    """Search the web for information"""

    return tavily_client.search(query)

web_search.invoke("Who is the current mayor of San Francisco?")
```

Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response.

agent = create_agent(
    model="gpt-5-nano",
    tools=[web_search]
)

question = HumanMessage(content="Who is the current mayor of San Francisco?")

response = agent.invoke(
    {"messages": [question]}
)
```
