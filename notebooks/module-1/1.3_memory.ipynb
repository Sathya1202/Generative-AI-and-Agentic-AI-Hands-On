## Key learnings

- Prompt structure and typed schemas make behavior more predictable, but do not guarantee correctness.
- Short-term memory requires both a checkpointer and a stable conversation/thread ID.

### Implementation step 8

Imports the libraries and APIs needed by the following example: from langgraph.checkpoint.memory import InMemorySaver. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Enables checkpoint-backed short-term memory. Reuse the same thread identifier to continue a conversation; a new ID starts a separate thread.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: from langgraph.checkpoint.memory import InMemorySaver. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Enables checkpoint-backed short-term memory. Reuse the same thread identifier to continue a conversation; a new ID starts a separate thread.

from langgraph.checkpoint.memory import InMemorySaver  


agent = create_agent(
    "gpt-5-nano",
    checkpointer=InMemorySaver(),  
)
```
