## Streaming Output

### Implementation step 22

Streams intermediate chunks/events instead of waiting for one complete response; this is useful for responsive user interfaces. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response. Inspects or prints the result so you can verify the output structure and follow the execution path.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Streams intermediate chunks/events instead of waiting for one complete response; this is useful for responsive user interfaces. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response. Inspects or prints the result so you can verify the output structure and follow the execution path.

for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
