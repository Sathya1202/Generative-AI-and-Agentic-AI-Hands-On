## Structured output

### Implementation step 10

Imports the libraries and APIs needed by the following example: from langchain.agents import create_agent; from langchain.messages import HumanMessage; from pydantic import BaseModel. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response. Inspects or prints the result so you can verify the output structure and follow the execution path.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: from langchain.agents import create_agent; from langchain.messages import HumanMessage; from pydantic import BaseModel. Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response. Inspects or prints the result so you can verify the output structure and follow the execution path.

from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

agent = create_agent(
    model='gpt-5-nano',
    system_prompt="You are a science fiction writer, create a capital city at the users request.",
    response_format=CapitalInfo
)

question = HumanMessage(content="What is the capital of The Moon?")

response = agent.invoke(
    {"messages": [question]}
)

response["structured_response"]
```
