### Implementation step 9

Imports the libraries and APIs needed by the following example: import base64. Prepares media bytes for a multimodal request. The MIME type must match the actual media format, and large media payloads may need size limits.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: import base64. Prepares media bytes for a multimodal request. The MIME type must match the actual media format, and large media payloads may need size limits.

import base64

# Get the first (and only) uploaded file dict
uploaded_file = uploader.value[0]

# This is a memoryview
content_mv = uploaded_file["content"]

# Convert memoryview -> bytes
img_bytes = bytes(content_mv)  # or content_mv.tobytes()

# Now base64 encode
img_b64 = base64.b64encode(img_bytes).decode("utf-8")
```

## Audio input

### Implementation step 12

Imports the libraries and APIs needed by the following example: import sounddevice as sd; from scipy.io.wavfile import write; import base64 (and 3 more import statement(s)) Prepares media bytes for a multimodal request. The MIME type must match the actual media format, and large media payloads may need size limits. Inspects or prints the result so you can verify the output structure and follow the execution path.

**What to verify:** Check that the output matches the intended data flow, and inspect intermediate messages/tool calls when behavior is unclear.


```python
# Purpose: Imports the libraries and APIs needed by the following example: import sounddevice as sd; from scipy.io.wavfile import write; import base64 (and 3 more import statement(s)) Prepares media bytes for a multimodal request. The MIME type must match the actual media format, and large media payloads may need size limits. Inspects or prints the result so you can verify the output structure and follow the execution path.

import sounddevice as sd
from scipy.io.wavfile import write
import base64
import io
import time
from tqdm import tqdm

# Recording settings
duration = 5  # seconds
sample_rate = 44100

print("Recording...")
audio = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1)
# Progress bar for the duration
for _ in tqdm(range(duration * 10)):   # update 10× per second
    time.sleep(0.1)
sd.wait()
print("Done.")

# Write WAV to an in-memory buffer
buf = io.BytesIO()
write(buf, sample_rate, audio)
wav_bytes = buf.getvalue()

aud_b64 = base64.b64encode(wav_bytes).decode("utf-8")
```

```python
# Purpose: Configures a chat model or agent. Model choice, system instructions and attached tools determine the capabilities available in this example. Executes the model/agent with an input payload. Inspect the returned object to understand messages, structured data, tool activity or state changes. Builds a typed conversation message. Message roles and ordering provide the context the model uses for its next response. Prepares media bytes for a multimodal request. The MIME type must match the actual media format, and large media payloads may need size limits. Inspects or prints the result so you can verify the output structure and follow the execution path.

agent = create_agent(
    model='gpt-audio',
)

multimodal_question = HumanMessage(content=[
    {"type": "text", "text": "Tell me about this audio file"},
    {"type": "audio", "base64": aud_b64, "mime_type": "audio/wav"}
])

response = agent.invoke(
    {"messages": [multimodal_question]}
)

print(response['messages'][-1].content)
```
